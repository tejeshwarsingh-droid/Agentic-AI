```python
import time 
import asyncio
```

```python
'''This is how the synchronous program works, one by one execution, 
    so if the program is big or large then we need to shift to asynchronous which do parallel execution
'''

def fetch_data():
    time.sleep(4)
    print("Data has been fetched")

def insert_data():
    time.sleep(2)
    print("Data has been added in the production")

def main():

    start_time = time.time()
    fetch_data()
    insert_data()
    end_time = time.time()
    print(f"Total time taken: {end_time-start_time}")
```

```python
main()
```

```python
''' Asynchronous program'''

async def fetch_dataa():
    await asyncio.sleep(4)
    print("Data has been fetched")

async def insert_dataa():
    await asyncio.sleep(2)
    print("Data has been added in the production")

async def main1():

    start_time = time.time()
    await asyncio.gather(fetch_dataa(),insert_dataa())

    end_time = time.time()
    print(f"Total time taken: {end_time-start_time}")
```

```python
await main1()
```
